# Employee Attrition EDA
This notebook loads the validated HR data, summarizes the target and leave patterns, and saves figures under `reports/figures/`.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
import seaborn as sns
sys.path.insert(0, str(Path.cwd().parent))
from src.data_loader import load_data
from src.config import FIGURES_DIR, TARGET
df = load_data()
df.head()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.countplot(data=df, x=TARGET, ax=axes[0])
sns.barplot(data=df, x='department', y=TARGET, ax=axes[1], errorbar=None)
sns.boxplot(data=df, x=TARGET, y='tenure_years', ax=axes[2])
axes[1].tick_params(axis='x', rotation=35)
fig.tight_layout()
fig.savefig(FIGURES_DIR / 'eda_target_department_tenure.png', dpi=140)
```

```python
leave_cols = ['leaves_taken_last_quarter', 'unplanned_leave_ratio', 'days_since_last_leave', TARGET]
sns.pairplot(df[leave_cols], hue=TARGET, corner=True)
plt.savefig(FIGURES_DIR / 'eda_leave_patterns.png', dpi=140)
```
